```python
#outliers summary
outlier_summary=[]
# outlier_df=[]
for col in numeric_cols:
    Q1=df[col].quantile(.25)
    Q3=df[col].quantile(.75)

    IQR=Q3-Q1
    upper_limit=Q3+1.5*(IQR)
    lower_limit=Q1-1.5*(IQR)

    outlier=df[(df[col]>upper_limit)|(df[col]<lower_limit)]
    outlier_summary.append({
         "column":col,
         "upper_limit":upper_limit,
         "lower_limit":lower_limit,
         "outlier_count":len(outlier)
     })
outlier_df=pd.DataFrame(outlier_summary)
print(outlier_df)
```

```python
#Feature Engineering for inventory Analysis
```

```python
#1.Inventory turnover proxy
df['Inventory_Turnover_proxy']=(df['Number of products sold']/df['Stock levels'].replace(0,np.nan))
k=(df['Number of products sold']/df['Stock levels'].replace(0,np.nan))
k
```

```python
#2.revenue per unit
df['Revenue_per_unit']=(df['Revenue generated']/df['Number of products sold'].replace(0,np.nan))
```

```python
#Production_sales_ratio
df['Production_sales_ratio']=(df['Production volumes']/df['Number of products sold'].replace(0,np.nan))
```

```python
df['Manufacturing_cost_per_unit']=(df['Manufacturing costs']/df['Production volumes'].replace(0,np.nan))
```

```python
#create stock categories
df['Stock_status']=pd.qcut(df['Stock levels'],q=3,labels=['Low','Medium','High'],duplicates='drop')
```

```python
df.head(2)
```

```python
#for supplieranalysis
#total suuplier-related cost
df['total_supply_cost']=(df['Shipping costs']+df['Manufacturing costs'])
#total lead time(total time from starting process to completing it)
df['total_lead_time']=(df['Manufacturing lead time']+df['Lead times'])
#Production efficiency
df['Production_efficiency']=(df["Production volumes"]/df['Manufacturing lead time'].replace(0,np.nan))
#defect category
df['Defect_category']=pd.cut(df["Defect rates"],bins=[-np.inf,1,3,np.inf],labels=['Low','Medium','High'])
```

```python
df.head(2)
```

```python
# df.rename(columns={"Manufacturing costs":"total_production_cost"},inplace=True)
```

```python
df['total_production_cost']=(df['Manufacturing costs']*df['Production volumes'])
```

```python
#total logistic cost
df['total_logistic_cost']=(df['Shipping costs']+df['Costs'])
```

```python
#shipping efficiency
df['Shipping_efficiency']=pd.cut(df["Shipping times"],bins=[-np.inf,2,5,np.inf],labels=['Fast','Moderate','Slow'])
```

```python
#supplier is with shorter or longer lead time
df['supplier_lead_time_category']=pd.qcut(df["Lead time"],q=3,labels=['Short','Medium','Long'])
```
